# Real Estate Recommender — Exploratory Data Analysis
Loads the three datasets, merges them, and explores the relationship between listing price and house size.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load datasets
users_df = pd.read_csv('data-refined/users.csv')
listings_df = pd.read_csv('data-refined/cleaned_listings.csv')
user_likes_df = pd.read_csv('data-refined/user_likes.csv')

print("Users Dataset:")
print(f"Shape: {users_df.shape}")
print(users_df.head())
print("\n" + "="*50 + "\n")

print("Listings Dataset:")
print(f"Shape: {listings_df.shape}")
print(listings_df.head())
print("\n" + "="*50 + "\n")

print("User Likes Dataset:")
print(f"Shape: {user_likes_df.shape}")
print(user_likes_df.head())
```

## Combined dataset
Join users, likes, and listing attributes into a single frame for quick inspection/feature work.

```python
# Merge users, likes, and listings into one dataframe
combined_df = (
    user_likes_df
    .merge(users_df, on='user_id', how='left')
    .merge(listings_df, on='listing_id', how='left')
)

print(f"Combined shape: {combined_df.shape}")
combined_df.head()
```

## Exploratory Data Analysis
Analyze the structure and relationships within the cleaned_listings dataset.

```python
# Scatterplot: Price vs Square Feet with trend line
plt.figure(figsize=(14, 6))
sns.scatterplot(data=listings_df, x='house_size', y='price', alpha=0.6, s=50)
sns.regplot(data=listings_df, x='house_size', y='price', scatter=False, color='red', line_kws={'linewidth': 2})
plt.title('Price vs Square Feet (with Trend Line)', fontsize=14, fontweight='bold')
plt.xlabel('Square Feet (sqft)', fontsize=12)
plt.ylabel('Price ($)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nPrice vs Sqft Correlation: {listings_df['price'].corr(listings_df['house_size']):.4f}")
```

```python
# Interaction sparsity
n_users = users_df['user_id'].nunique()
n_items = listings_df['listing_id'].nunique()
n_interactions = len(user_likes_df)
sparsity = 1 - n_interactions / (n_users * n_items)

print(f"Users:        {n_users}")
print(f"Listings:     {n_items}")
print(f"Interactions: {n_interactions}")
print(f"Sparsity:     {sparsity:.4%}")
print(f"Avg likes per user: {n_interactions / n_users:.1f}")
```
